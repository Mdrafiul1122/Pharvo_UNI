```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

driver = webdriver.Chrome()
driver.maximize_window()
wait = WebDriverWait(driver, 10)
```

```python
try:
    driver.get("http://localhost:5173/")
    driver.execute_script("window.localStorage.clear(); window.sessionStorage.clear();")
    driver.get("http://localhost:5173/login")
    wait.until(EC.presence_of_element_located((By.ID, "username")))

    # Safe, repeatable success feedback that changes no data (verified in login.jsx):
    # the Demo Login "Copy" button flips its label to "Copied!" for 2 seconds.
    copy_btn = wait.until(EC.element_to_be_clickable((By.XPATH, "//button[contains(@class, 'demo-login__copy')]")))
    assert copy_btn.text.strip() == "Copy", f"Unexpected initial label: {copy_btn.text!r}"
    copy_btn.click()
    wait.until(lambda d: d.find_element(By.XPATH, "//button[contains(@class, 'demo-login__copy')]").text.strip() == "Copied!")
    print("Success message shown: Copied!")
    print("PASS: Success message verified")
except Exception as e:
    print("FAIL:", e)
    driver.save_screenshot("54_success_message_FAIL.png")
finally:
    driver.quit()
```
