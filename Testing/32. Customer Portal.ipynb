```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

driver = webdriver.Chrome()
driver.maximize_window()
wait = WebDriverWait(driver, 10)

driver.get("http://localhost:5173/")
driver.execute_script("window.localStorage.clear(); window.sessionStorage.clear();")
driver.get("http://localhost:5173/login")

wait.until(EC.presence_of_element_located((By.ID, "username")))
driver.find_element(By.ID, "username").send_keys("shakil.ahmed@gmail.com")
driver.find_element(By.ID, "password").send_keys("12345678")
driver.find_element(By.ID, "sign-in-btn").click()

time.sleep(3)
print("URL:", driver.current_url)
print("Page text:", driver.find_element(By.TAG_NAME, "body").text[:300])
```

```python
try:
    # Real portal nav (verified: nav[aria-label='Customer portal'] in PortalChrome.jsx)
    nav = wait.until(EC.visibility_of_element_located((By.XPATH, "//nav[@aria-label='Customer portal']")))
    nav_text = nav.text
    for item in ["Home", "Purchases", "Benefits", "Profile"]:
        assert item in nav_text, f"Portal menu item missing: {item}"
    print("Portal menu visible: Home, Purchases, Benefits, Profile")

    # Must NOT be the staff shell (verified: nav[aria-label='Staff'] in AppShell.jsx)
    assert not driver.find_elements(By.XPATH, "//nav[@aria-label='Staff']"), \
        "Staff shell shown to a customer account."
    print("Staff shell correctly absent.")

    print("Current URL:", driver.current_url)
    print("Page text:", driver.find_element(By.TAG_NAME, "body").text[:300])
    print("PASS: Customer Portal")
except Exception as e:
    print("FAIL: Customer Portal")
    print("Error:", e)
    driver.save_screenshot("32_customer_portal_FAIL.png")
```
