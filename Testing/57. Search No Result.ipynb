```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

driver = webdriver.Chrome()
driver.maximize_window()
wait = WebDriverWait(driver, 10)
```

```python
try:
    driver.get("http://localhost:5173/")
    driver.execute_script("window.localStorage.clear(); window.sessionStorage.clear();")
    driver.get("http://localhost:5173/login")
    wait.until(EC.presence_of_element_located((By.ID, "username")))
    driver.find_element(By.ID, "username").send_keys("shawon@gmail.com")
    driver.find_element(By.ID, "password").send_keys("12345678")
    driver.find_element(By.ID, "sign-in-btn").click()
    time.sleep(3)
    wait.until(EC.visibility_of_element_located((By.XPATH, "//nav[@aria-label='Staff']")))

    # Real search field (verified aria-label in CustomerDirectory.jsx).
    # Search is debounced (350ms) server-side (verified in CustomersPage.jsx).
    wait.until(EC.element_to_be_clickable((By.XPATH, "//aside//button[contains(., 'Customers')]"))).click()
    search = wait.until(EC.visibility_of_element_located((By.XPATH, "//input[@aria-label='Search customers by name, phone or email']")))
    search.clear()
    search.send_keys("Selenium_No_Result_987654")
    time.sleep(3)

    # Real empty state (verified text in CustomerDirectory.jsx)
    body = driver.find_element(By.TAG_NAME, "body").text
    assert "No customers found" in body, f"No-result state not shown. Page: {body[:200]}"
    print("Search term: Selenium_No_Result_987654")
    print("No-result state shown: No customers found")
    print("PASS: Search no-result behavior verified")
except Exception as e:
    print("FAIL:", e)
    driver.save_screenshot("57_search_no_result_FAIL.png")
finally:
    driver.quit()
```
